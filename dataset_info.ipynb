```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("tweet_emotions.csv")
```

```python
df.sentiment.unique()
```

```python
emotion_counts = df["sentiment"].value_counts()
total_tweets = len(df)
emotion_percentages = (emotion_counts / total_tweets) * 100
```

```python
distribution_table = pd.DataFrame({
    "Συναίσθημα": emotion_counts.index,
    "Πλήθος": emotion_counts.values,
    "Ποσοστό (%)": emotion_percentages.values
})

print(distribution_table)
```

```python
plt.figure(figsize=(10, 6))
emotion_percentages.sort_values().plot(kind='barh', color='skyblue', edgecolor='black')
plt.xlabel("Ποσοστό (%)")
plt.ylabel("Συναίσθημα")
plt.title("Κατανομή Συναισθημάτων στο Σύνολο Δεδομένων")
plt.grid(axis="x", linestyle="--", alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 10))
explode = [0.1 if p < 3 else 0 for p in emotion_percentages]  # Απομακρύνει κατηγορίες με ποσοστό < 3%
plt.pie(emotion_percentages, labels=emotion_counts.index, autopct=lambda p: f'{p:.1f}%' if p > 1 else '', 
        colors=plt.cm.Paired.colors, startangle=140, wedgeprops={'edgecolor': 'black'})

plt.title("Κατανομή Συναισθημάτων στο Σύνολο Δεδομένων", pad=30)
plt.axis("equal")  # Διασφάλιση ότι ο κύκλος παραμένει κυκλικός
plt.show()
```
